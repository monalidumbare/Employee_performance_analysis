```python
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Image,
    Table,
    TableStyle,
    PageBreak
)

from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib import colors
from reportlab.lib.units import inch
import os

# Create reports folder
os.makedirs("reports", exist_ok=True)

# PDF file path
pdf_file = "reports/Employee_HR_Analytics_Report.pdf"

# Create PDF document
doc = SimpleDocTemplate(pdf_file)

styles = getSampleStyleSheet()
content = []

# --------------------------------
# 1. PROJECT TITLE
# --------------------------------

content.append(
    Paragraph(
        "Employee Performance & HR Analytics",
        styles["Title"]
    )
)

content.append(Spacer(1, 20))

# --------------------------------
# 2. DATASET OVERVIEW
# --------------------------------

content.append(
    Paragraph("1. Dataset Overview", styles["Heading1"])
)

overview = f"""
Total Employees: {len(df)}<br/>
Total Columns: {len(df.columns)}<br/>
Total Departments: {df['Department'].nunique()}<br/>
Average Monthly Salary: {df['MonthlyIncome'].mean():.2f}<br/>
Average Age: {df['Age'].mean():.2f}<br/>
Average Experience: {df['TotalWorkingYears'].mean():.2f} years
"""

content.append(Paragraph(overview, styles["BodyText"]))
content.append(Spacer(1, 20))

# --------------------------------
# 3. DATA CLEANING SUMMARY
# --------------------------------

content.append(
    Paragraph("2. Data Cleaning Summary", styles["Heading1"])
)

cleaning = f"""
Missing Values: {df.isnull().sum().sum()}<br/>
Duplicate Records: {df.duplicated().sum()}<br/>
Total Records: {len(df)}<br/>
Total Columns: {len(df.columns)}
"""

content.append(Paragraph(cleaning, styles["BodyText"]))
content.append(Spacer(1, 20))

# --------------------------------
# 4. STATISTICAL SUMMARY
# --------------------------------

content.append(
    Paragraph("3. Statistical Summary", styles["Heading1"])
)

stats = df[
    ["Age", "MonthlyIncome", "TotalWorkingYears"]
].describe()

stats_data = [
    ["Statistic"] + list(stats.columns)
]

for index, row in stats.iterrows():
    stats_data.append(
        [index] + [round(value, 2) for value in row]
    )

table = Table(stats_data)

table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.darkblue),
    ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
    ("GRID", (0, 0), (-1, -1), 1, colors.grey),
    ("ALIGN", (0, 0), (-1, -1), "CENTER"),
    ("PADDING", (0, 0), (-1, -1), 8)
]))

content.append(table)
content.append(PageBreak())

# --------------------------------
# 5. ADD ALL 12 VISUALIZATIONS
# --------------------------------

content.append(
    Paragraph("4. Data Visualizations", styles["Heading1"])
)

chart_files = [
    "salary_distribution.png",
    "age_distribution.png",
    "salary_boxplot.png",
    "experience_boxplot.png",
    "department_count.png",
    "gender_count.png",
    "education_count.png",
    "average_salary_department.png",
    "business_travel.png",
    "experience_salary.png",
    "correlation_heatmap.png",
    "pairplot.png"
]

for i, chart in enumerate(chart_files):

    path = f"reports/visualizations/{chart}"

    if os.path.exists(path):

        content.append(
            Paragraph(
                chart.replace(".png", "").replace("_", " ").title(),
                styles["Heading2"]
            )
        )

        content.append(
            Image(
                path,
                width=6.2 * inch,
                height=3.7 * inch
            )
        )

        content.append(Spacer(1, 15))

        if i < len(chart_files) - 1:
            content.append(PageBreak())

# --------------------------------
# 6. BUSINESS INSIGHTS
# --------------------------------

content.append(PageBreak())

content.append(
    Paragraph("5. Business Insights", styles["Heading1"])
)

department_count = df["Department"].value_counts()

highest_salary_department = (
    df.groupby("Department")["MonthlyIncome"].mean().idxmax()
)

attrition_department = (
    df[df["Attrition"] == "Yes"]["Department"].value_counts()
)

attrition_rate = (
    df["Attrition"].eq("Yes").mean() * 100
)

insights = [
    f"Department with highest employee count: {department_count.idxmax()}.",

    f"Department with highest average salary: {highest_salary_department}.",

    f"Department with highest attrition count: {attrition_department.idxmax()}.",

    f"Overall employee attrition rate: {attrition_rate:.2f}%.",

    f"Average monthly salary: {df['MonthlyIncome'].mean():.2f}.",

    f"Average employee experience: {df['TotalWorkingYears'].mean():.2f} years."
]

for insight in insights:

    content.append(
        Paragraph("• " + insight, styles["BodyText"])
    )

    content.append(Spacer(1, 12))

# --------------------------------
# 7. CONCLUSION
# --------------------------------

content.append(
    Paragraph("6. Conclusion", styles["Heading1"])
)

conclusion = """
This project analyzes employee demographics, salary distribution,
department performance, employee experience and attrition.
The analysis provides useful information to help HR teams
understand workforce trends and make data-driven decisions.
"""

content.append(
    Paragraph(conclusion, styles["BodyText"])
)

# --------------------------------
# 8. GENERATE PDF
# --------------------------------

doc.build(content)

print("PDF Report Generated Successfully!")
print("File Location:", os.path.abspath(pdf_file))
```

```python

import os
import pandas as pd

from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    Image,
    Table,
    TableStyle,
    PageBreak
)

from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors
from reportlab.lib.units import inch
from reportlab.lib.enums import TA_CENTER
from PIL import Image as PILImage

# ---------------------------------------
# 1. CREATE REPORT FOLDER
# ---------------------------------------

os.makedirs("reports", exist_ok=True)

pdf_file = "reports/Employee_HR_Analytics_Report.pdf"

# ---------------------------------------
# 2. CREATE PDF DOCUMENT
# ---------------------------------------

doc = SimpleDocTemplate(
    pdf_file,
    pagesize=(8.27 * inch, 11.69 * inch),
    rightMargin=0.65 * inch,
    leftMargin=0.65 * inch,
    topMargin=0.65 * inch,
    bottomMargin=0.65 * inch
)

styles = getSampleStyleSheet()
content = []

# Custom styles
title_style = ParagraphStyle(
    "CustomTitle",
    parent=styles["Title"],
    fontSize=22,
    textColor=colors.HexColor("#17365D"),
    alignment=TA_CENTER,
    spaceAfter=20
)

heading_style = ParagraphStyle(
    "CustomHeading",
    parent=styles["Heading1"],
    textColor=colors.HexColor("#087E8B"),
    spaceBefore=12,
    spaceAfter=12
)

# ---------------------------------------
# 3. PROJECT TITLE
# ---------------------------------------

content.append(
    Paragraph(
        "EMPLOYEE PERFORMANCE & HR ANALYTICS",
        title_style
    )
)

content.append(
    Paragraph(
        "Data Analysis and Business Intelligence Report",
        styles["Heading2"]
    )
)

content.append(Spacer(1, 20))

# ---------------------------------------
# 4. DATASET OVERVIEW
# ---------------------------------------

content.append(
    Paragraph("1. Dataset Overview", heading_style)
)

overview_data = [
    ["Metric", "Value"],
    ["Total Employees", str(len(df))],
    ["Total Columns", str(len(df.columns))],
    ["Total Departments", str(df["Department"].nunique())],
    ["Average Age", f"{df['Age'].mean():.2f} years"],
    ["Average Monthly Salary", f"{df['MonthlyIncome'].mean():,.2f}"],
    ["Average Experience", f"{df['TotalWorkingYears'].mean():.2f} years"]
]

overview_table = Table(
    overview_data,
    colWidths=[3 * inch, 3 * inch]
)

overview_table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#17365D")),
    ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
    ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
    ("BACKGROUND", (0, 1), (-1, -1), colors.HexColor("#EAF2F8")),
    ("PADDING", (0, 0), (-1, -1), 8)
]))

content.append(overview_table)
content.append(Spacer(1, 20))

# ---------------------------------------
# 5. DATA CLEANING SUMMARY
# ---------------------------------------

content.append(
    Paragraph("2. Data Cleaning Summary", heading_style)
)

cleaning_text = f"""
Total Records: {len(df)}<br/>
Missing Values Remaining: {df.isnull().sum().sum()}<br/>
Duplicate Records Remaining: {df.duplicated().sum()}<br/>
Total Columns: {len(df.columns)}<br/>
Dataset Shape: {df.shape}
"""

content.append(
    Paragraph(cleaning_text, styles["BodyText"])
)

content.append(Spacer(1, 20))

# ---------------------------------------
# 6. STATISTICAL SUMMARY
# ---------------------------------------

content.append(
    Paragraph("3. Statistical Summary", heading_style)
)

stats = df[
    ["Age", "MonthlyIncome", "TotalWorkingYears"]
].describe().round(2)

stats_data = [
    ["Statistic"] + list(stats.columns)
]

for index, row in stats.iterrows():
    stats_data.append(
        [str(index)] + [str(value) for value in row]
    )

stats_table = Table(stats_data, repeatRows=1)

stats_table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#087E8B")),
    ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
    ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
    ("ALIGN", (0, 0), (-1, -1), "CENTER"),
    ("PADDING", (0, 0), (-1, -1), 7)
]))

content.append(stats_table)

content.append(PageBreak())

# ---------------------------------------
# 7. ADD 12 COLORFUL VISUALIZATIONS
# ---------------------------------------

content.append(
    Paragraph("4. Data Visualizations", heading_style)
)

chart_files = [
    "salary_distribution.png",
    "age_distribution.png",
    "salary_boxplot.png",
    "experience_boxplot.png",
    "department_count.png",
    "gender_count.png",
    "education_count.png",
    "average_salary_department.png",
    "business_travel.png",
    "experience_salary.png",
    "correlation_heatmap.png",
    "pairplot.png"
]

chart_folder = "reports/visualizations"

for i, chart in enumerate(chart_files):

    chart_location = os.path.join(chart_folder, chart)

    if not os.path.exists(chart_location):
        raise FileNotFoundError(
            f"Chart not found: {chart_location}"
        )

    content.append(
        Paragraph(
            f"{i + 1}. {chart.replace('.png', '').replace('_', ' ').title()}",
            heading_style
        )
    )

    # Preserve original image proportions
    with PILImage.open(chart_location) as img:
        img_width, img_height = img.size

    max_width = 6.7 * inch
    max_height = 5.3 * inch

    scale = min(
        max_width / img_width,
        max_height / img_height
    )

    report_image = Image(
        chart_location,
        width=img_width * scale,
        height=img_height * scale
    )

    content.append(report_image)
    content.append(Spacer(1, 12))

    if i < len(chart_files) - 1:
        content.append(PageBreak())

# ---------------------------------------
# 8. BUSINESS INSIGHTS
# ---------------------------------------

content.append(PageBreak())

content.append(
    Paragraph("5. Business Insights", heading_style)
)

department_count = df["Department"].value_counts()

salary_by_department = (
    df.groupby("Department")["MonthlyIncome"].mean()
)

attrition_by_department = (
    df[df["Attrition"] == "Yes"]["Department"].value_counts()
)

attrition_rate = (
    df["Attrition"].eq("Yes").mean() * 100
)

insights = [
    f"Department with highest employee count: {department_count.idxmax()}.",

    f"Department with highest average salary: {salary_by_department.idxmax()}.",

    f"Department with highest attrition count: "
    f"{attrition_by_department.idxmax() if len(attrition_by_department) else 'No attrition records'}.",

    f"Overall employee attrition rate: {attrition_rate:.2f}%.",

    f"Average monthly salary: {df['MonthlyIncome'].mean():,.2f}.",

    f"Average employee experience: {df['TotalWorkingYears'].mean():.2f} years."
]

for insight in insights:

    content.append(
        Paragraph(
            "&#8226; " + insight,
            styles["BodyText"]
        )
    )

    content.append(Spacer(1, 12))

# ---------------------------------------
# 9. CONCLUSION
# ---------------------------------------

content.append(
    Paragraph("6. Conclusion", heading_style)
)

conclusion = """
The Employee Performance and HR Analytics project provides
insights into employee demographics, salary distribution,
department performance, experience and employee attrition.

These findings can help HR teams improve workforce planning,
employee retention, salary management and performance monitoring.
"""

content.append(
    Paragraph(conclusion, styles["BodyText"])
)

# ---------------------------------------
# 10. GENERATE PDF
# ---------------------------------------

doc.build(content)

print("===================================")
print("PDF REPORT GENERATED SUCCESSFULLY!")
print("===================================")

print("PDF File:", pdf_file)

print(
    "File Size:",
    round(os.path.getsize(pdf_file) / 1024, 2),
    "KB"
)
```

```python

from docx import Document
from docx.shared import Pt, Inches, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
import os

# Create reports folder
os.makedirs("reports", exist_ok=True)

# Create document
word = Document()

# Set default font
style = word.styles["Normal"]
style.font.name = "Arial"
style.font.size = Pt(11)

# --------------------------------
# TITLE
# --------------------------------

title = word.add_heading(
    "Employee Performance & HR Analytics",
    0
)

title.alignment = WD_ALIGN_PARAGRAPH.CENTER

subtitle = word.add_paragraph(
    "Business Insights and Recommendations"
)
subtitle.alignment = WD_ALIGN_PARAGRAPH.CENTER

word.add_paragraph(" ")

# --------------------------------
# 1. PROJECT OVERVIEW
# --------------------------------

word.add_heading("1. Project Overview", level=1)

word.add_paragraph(
    "The Employee Performance and HR Analytics project analyzes "
    "employee demographics, salary distribution, department "
    "productivity, education, experience, performance and attrition. "
    "The objective is to identify workforce trends and support "
    "better human resource decisions."
)

# --------------------------------
# 2. DATASET OVERVIEW
# --------------------------------

word.add_heading("2. Dataset Overview", level=1)

word.add_paragraph(f"Total Employees: {len(df)}")
word.add_paragraph(f"Total Columns: {len(df.columns)}")
word.add_paragraph(f"Total Departments: {df['Department'].nunique()}")
word.add_paragraph(
    f"Average Monthly Salary: {df['MonthlyIncome'].mean():,.2f}"
)

# --------------------------------
# 3. KEY BUSINESS INSIGHTS
# --------------------------------

word.add_heading("3. Key Business Insights", level=1)

department_count = df["Department"].value_counts()

salary_by_department = (
    df.groupby("Department")["MonthlyIncome"].mean()
)

attrition_by_department = (
    df[df["Attrition"] == "Yes"]["Department"].value_counts()
)

attrition_rate = df["Attrition"].eq("Yes").mean() * 100

insights = [
    f"{department_count.idxmax()} has the highest employee count.",

    f"{salary_by_department.idxmax()} has the highest average salary.",

    f"Overall employee attrition rate is {attrition_rate:.2f}%.",

    f"{attrition_by_department.idxmax()} has the highest attrition count."
    if len(attrition_by_department)
    else "No employee attrition records were found.",

    f"Average employee experience is "
    f"{df['TotalWorkingYears'].mean():.2f} years.",

    f"Average employee age is {df['Age'].mean():.2f} years."
]

for insight in insights:
    word.add_paragraph(insight, style="List Bullet")

# --------------------------------
# 4. HR PROBLEMS IDENTIFIED
# --------------------------------

word.add_heading("4. HR Problems Identified", level=1)

problems = [
    "Employee attrition can affect workforce stability.",
    "Salary distribution may vary across departments.",
    "Employee performance requires regular monitoring.",
    "Career growth and promotion opportunities should be evaluated.",
    "Employee satisfaction should be monitored to improve retention."
]

for problem in problems:
    word.add_paragraph(problem, style="List Bullet")

# --------------------------------
# 5. BUSINESS RECOMMENDATIONS
# --------------------------------

word.add_heading("5. Business Recommendations", level=1)

recommendations = [
    "Develop effective employee retention strategies.",
    "Review salary distribution across departments.",
    "Provide training and development opportunities.",
    "Monitor employee performance regularly.",
    "Improve career growth and promotion opportunities.",
    "Analyze overtime and employee satisfaction."
]

for recommendation in recommendations:
    word.add_paragraph(recommendation, style="List Bullet")

# --------------------------------
# 6. CONCLUSION
# --------------------------------

word.add_heading("6. Conclusion", level=1)

word.add_paragraph(
    "HR Analytics helps organizations understand workforce trends "
    "and employee behavior. The analysis provides useful insights "
    "for employee retention, salary management, performance "
    "improvement and workforce planning."
)

# --------------------------------
# SAVE DOCUMENT
# --------------------------------

file_path = "reports/HR_Business_Insights.docx"

word.save(file_path)

print("Business Insights Document Created Successfully!")
print("File Location:", os.path.abspath(file_path))
```
